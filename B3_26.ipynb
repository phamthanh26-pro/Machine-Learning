```python
def function (x):
  return x**2 - 4*x +5
```

```python
def derivative(x):
  return 2*x - 4
```

```python
def myGD(x0, eta):
    x = [x0]
    for it in range(100):
        x_new = x[-1] - eta * derivative(x[-1])
        x.append(x_new)
        # Điều kiện dừng khi đạo hàm rất nhỏ (gần cực trị)
        if abs(derivative(x_new)) < 1e-3:
            break
    return (x, it)
```

```python
(x1, it1) = myGD(5, 0.2)
print ('Đáp án x1 = %f,cost = %f,after %d iteration ' % (x1[-1], derivative(x1[-1]), it1))
```
