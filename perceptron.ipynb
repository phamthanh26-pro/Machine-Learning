```python
import numpy as np

# Dữ liệu của bài toán (x đã thêm bias = 1 ở cuối)
w = np.array([1, 2, -10])
x = np.array([3, 4, 1])
y = -1                      # nhãn thật

def du_doan(w, x):

    if np.dot(w, x) >= 0:
        return 1
    return -1

# Bước 1: tính w^T x
print("w^T x =", np.dot(w, x))

# Bước 2: dự đoán nhãn
nhan_du_doan = du_doan(w, x)
print("Nhãn dự đoán =", nhan_du_doan)

# Bước 3: kiểm tra đúng/sai
if nhan_du_doan != y:
    print("Phân lớp SAI -> cập nhật w")
    w = w + y * x
else:
    print("Phân lớp ĐÚNG -> giữ nguyên w")

# Kiểm tra lại sau khi cập nhật
print("Dự đoán mới =", du_doan(w, x))    # -1 (đúng)
```
